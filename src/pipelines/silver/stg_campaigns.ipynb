```python
CATALOG = "medallion_catalog" 
BRONZE_SCHEMA = "bronze"
SILVER_SCHEMA = "silver"
```

```python
spark.sql(f"""

CREATE OR REPLACE TABLE {CATALOG}.{SILVER_SCHEMA}.stg_campaigns


WITH bronze_source AS (
    SELECT * 
    FROM {CATALOG}.{BRONZE_SCHEMA}.campaigns
)


SELECT  campaign_id
        , campaign_name
        , channel as campaign_channel
        , start_date as campaign_start_date
        , end_date as campaign_end_date
        , budget as campaign_budget
        , actual_spend as campaign_actual_spend
FROM bronze_source
    
"""
)
```

```python
spark.sql(f"""
    COMMENT ON TABLE {CATALOG}.{SILVER_SCHEMA}.stg_campaigns IS 
    "Staging table for campaigns"
""")

descriptions = {
    "campaign_id": "Unique identifier for the campaign",
    "campaign_name": "Name of the marketing campaign",
    "campaign_channel": "Marketing channel used for the campaign (e.g., email, social, search)",
    "campaign_start_date": "Date the campaign started",
    "campaign_end_date": "Date the campaign ended",
    "campaign_budget": "Planned budget allocated for the campaign in USD",
    "campaign_actual_spend": "Actual amount spent on the campaign in USD",
}

for col, desc in descriptions.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SILVER_SCHEMA}.stg_campaigns ALTER COLUMN {col} COMMENT '{desc}'")

spark.sql(f"ALTER TABLE {CATALOG}.{SILVER_SCHEMA}.stg_campaigns ALTER COLUMN campaign_id SET NOT NULL")
spark.sql(f"ALTER TABLE {CATALOG}.{SILVER_SCHEMA}.stg_campaigns ADD CONSTRAINT campaign_pk PRIMARY KEY (campaign_id)")
```
